```python
def isValid(s: str) -> bool:
        parentheses = {"(":")","[":"]","{":"}"}
        stack = []
        if len(s) % 2 != 0:
            return False
        for parenthese in s:
            if parenthese in parentheses:
                stack.append(parenthese)
            elif not stack or parentheses[stack.pop()] != parenthese:
                return False
        return True

assert isValid("()") == True
assert isValid("()[]{}") == True
assert isValid("(]") == False
assert isValid("([])") == True
assert isValid("([)]") == False
print("Test passed")
```
