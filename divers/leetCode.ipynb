### 1. Two Sum

```python
def two_sum(nums: list[int], target: int):
    for i, j in enumerate(nums):
        for k, l in enumerate(nums):
            if j + l == target and i != k:
                return [i, k]

assert two_sum([2,7,11,15], 9) == [0,1]
assert two_sum([3,2,4], 6) == [1,2]
assert two_sum([3,3], 6) == [0,1]
print("Tests passed")
```

### 14. Longest Common Prefix

```python
def longestCommonPrefix(strs: list[str]) -> str:
    res = ""
    for c in zip(*strs):
        if len(set(c)) == 1:
            res += c[0]
        else:
            break
    return res

assert longestCommonPrefix(["flower", "flow", "flight"]) == "fl"
assert longestCommonPrefix(["cir", "car"]) == "c"
print("Tests passed")
```

### 20. Valid Parentheses

```python
def isValid(s: str) -> bool:
        parentheses = {"(":")","[":"]","{":"}"}
        stack = []
        if len(s) % 2 != 0:
            return False
        for parenthese in s:
            if parenthese in parentheses:
                stack.append(parenthese)
            elif not stack or parentheses[stack.pop()] != parenthese:
                return False
        return not stack

assert isValid("()") == True
assert isValid("()[]{}") == True
assert isValid("(]") == False
assert isValid("([])") == True
assert isValid("([)]") == False
assert isValid("((") == False
print("Test passed")
```

### 66. Plus One

```python
def plusOne(digits: list[int]) -> list[int]:
        r = int(''.join(list(map(str, digits)))) + 1
        return [int(x) for x in str(r)]

assert plusOne([1,2,3]) == [1,2,4]
assert plusOne([4,3,2,1]) == [4,3,2,2]
print("Tests passed")
```

### 88. Merge Sorted Array

```python
def merge(nums1: list[int], m: int, nums2: list[int], n: int) -> None:
        """
        Do not return anything, modify nums1 in-place instead.
        """
        r1 = m -1
        r2 = n -1
        w = m + n -1
        while r2 >= 0:
            if r1 >= 0 and nums1[r1] > nums2[r2]:
                nums1[w] = nums1[r1]
                r1 -= 1
            else:
                nums1[w] = nums2[r2]
                r2 -= 1
            w -= 1

# Test 1
nums1 = [1, 2, 3, 0, 0, 0]
merge(nums1, 3, [2, 5, 6], 3)
assert nums1 == [1, 2, 2, 3, 5, 6]

# Test 2
nums1 = [1]
merge(nums1, 1, [], 0)
assert nums1 == [1]

# Test 3
nums1 = [0]
merge(nums1, 0, [1], 1)
assert nums1 == [1]

print("Tests passed")
```

### 101. Symmetric Tree

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def isSymmetric(root: TreeNode | None) -> bool:
        if not root:
            return True
        def isMirror(t1: TreeNode | None, t2: TreeNode | None):
            if not t1 and not t2:
                return True
            if not t1 or not t2 or t1.val != t2.val:
                return False
            return isMirror(t1.left, t2.right) and isMirror(t1.right, t2.left)
        return isMirror(root.left, root.right)

# --- Préparation du test ---
# Construction manuelle de l'arbre représenté par [1, 2, 2, 3, 4, 4, 3]
#        1
#       / \
#      2   2
#     / \ / \
#    3  4 4  3

root = TreeNode(1)
root.left = TreeNode(2, TreeNode(3), TreeNode(4))
root.right = TreeNode(2, TreeNode(4), TreeNode(3))

# L'assertion passe maintenant à 100% (True est un booléen, pas la chaîne "true")
assert isSymmetric(root) == True
print("Test passed")
```

### 125. Valid Palindrome

```python
def isPalindrome(s: str):
    s = [x.lower() for x in s if x.isalnum()]
    return s == s[::-1]

assert isPalindrome("A man, a plan, a canal: Panama") == True
assert isPalindrome("race a car") == False
assert isPalindrome(" ") == True
print("Tests passed")
```

### 217. Contains Duplicate

```python
def containsDuplicate(nums: list[int]) -> bool:
    return len(nums) != len(set(nums))

assert containsDuplicate([1,2,3,1]) == True
assert containsDuplicate([1,2,3,4]) == False
assert containsDuplicate([1,1,1,3,3,4,3,2,4,2]) == True
print("Tests passed")
```

### 242. Valid Anagram

```python
def isAnagram(s: str, t: str) -> bool:
    return sorted(s) == sorted(t)

assert isAnagram("anagram", "nagaram") == True
assert isAnagram("rat", "car") == False
print("Tests passed")
```

### 392. Is Subsequence

```python
def isSubsequence(s: str, t: str) -> bool:
    p1, p2 = 0, 0
    while p2 < len(t):
        if s[p1] == t[p2]:
            p1 += 1
        p2 += 1


# assert isSubsequence("abc", "ahbgdc") == True
assert isSubsequence("axc", "ahbgdc") == False
print("Tests passed")
```
